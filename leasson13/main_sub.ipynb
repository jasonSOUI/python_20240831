```python
from datetime import date, datetime

import os
import csv

def create_csv_file(file_path, data):
    # 獲取檔案的目錄路徑
    directory = os.path.dirname(file_path)
    
    # 如果目錄不存在,則建立它
    if not os.path.exists(directory):
        os.makedirs(directory)
    
    # 開啟檔案並寫入資料
    with open(file_path, 'w', newline='', encoding='utf-8') as csvfile:
        writer = csv.writer(csvfile)
        writer.writerows(data)

# 範例使用
today = datetime.now()
data_path = os.getcwd()
data_dir = today.strftime("%Y%m%d")
data_file_path = os.path.join(data_path, data_dir) + ".csv"

data = [
    ['時間', '設備', '值']
]

d1 = []
d1.append(today.strftime("%Y-%m-%d %H:%M:%S"))
d1.append("溫度")
d1.append(23.1)
data.append(d1)

create_csv_file(data_file_path, data)
print(f"CSV 檔案已成功建立於: {data_file_path}")
```
